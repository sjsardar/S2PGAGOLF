### Bar chart showing scores in certain stroke types, would have a selectable drop down at the top to select which tournament we are looking at


sg_putt: strokes gained putting  `\n`
sg_arg: strokes gained around the green `\n`
sg_app: strokes gained approach  `\n`
sg_ott: strokes gained of the tee `\n`

```python
stroke_cols = ['sg_putt', 'sg_arg', 'sg_app', 'sg_ott']
tournaments = sorted(pga_raw['tournament name'].dropna().unique())

def plot_tournament(tournament, top_n=15):
    df = pga_raw[pga_raw['tournament name'] == tournament]

    data = (df.groupby('Player_initial_last')[stroke_cols + ['sg_total']].mean().sort_values('sg_total', ascending=False).head(top_n)[stroke_cols])

    fig, ax = plt.subplots(figsize=(10, 6), facecolor='#DAD7CD')
    data.plot(kind='bar', stacked=True, ax=ax)
    ax.set_title(f'Strokes Gained by Category: {tournament}')
    ax.set_ylabel('Average Strokes Gained per Round')
    ax.axhline(0, color='black', linewidth=0.8)
    ax.legend(title='Strokes Gained Category')
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig('Strokes Gained by Category.png', dpi=300, transparent=False)
    plt.show()

widgets.interact(
    plot_tournament,
    tournament=widgets.Dropdown(options=tournaments, description='Tournament:'),
    top_n=widgets.IntSlider(value=15, min=5, max=50, step=5, description='Top N:'),
);
```
